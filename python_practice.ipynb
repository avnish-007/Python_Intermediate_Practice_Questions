# 1. Check Whether a Number is Even or Odd
**Problem Statement**

Given an integer N, determine whether the given number is even or odd. A number is even if it is completely
divisible by 2. Otherwise, it is odd.

**Input:**

A single integer N.

**Output:**

```python
n = int(input("Enter the number:"))
if n % 2 == 0:
    print("The number is Even.")
else:
    print("The number is Odd.")
```

# 19. Check Whether a Number is Prime

**Problem Statement**

Given an integer N, determine whether it is a prime number. A prime number is greater than 1 and has
exactly two factors: 1 and itself.

**Input**

A single integer N.

**Output**

Print Prime or Not Prime.

```python
n = int(input().strip())

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

# 20. Print All Prime Numbers in a Given Range

**Problem Statement**

Given two integers L and R, print all prime numbers between them.

**Input**

Two integers L and R.

**Output**

Print all prime numbers in the range [L, R].

```python
l = int(input().strip())
r = int(input().strip())
for i in range(l,r+1):
    if i <=1:
        continue

    is_prime = True
    for j in range(2,int(i**0.5)+1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        print(i)
```

# 21. Find the Prime Factors of a Number

**Problem Statement**

**Input**

A single integer N.

**Output**

Print the prime factors, including repeated factors.

```python
n = int(input())

i = 2
while n > 1:
    if n % i == 0:
        print(i, end=" ")
        n //= i
    else:
        i += 1
```

# 22. Express a Number as the Sum of Two Prime Numbers

**Problem Statement**

Given an integer N, determine whether it can be expressed as the sum of two prime numbers.

**Input**

A single integer N.

**Output**

Print True if such a pair exists; otherwise print False.

```python
def IsPrime(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i == 0 :
            return False
    return True

def SumOfTwoPrime(n):
    if n < 4:
        return False
    elif n % 2 == 0:
        return True
    else:
        return IsPrime(n-2)

i = int(input().strip())
print(SumOfTwoPrime(i))
```

# 23. Check Whether a Number is an Armstrong Number

of each digit raised to the power of the number of digits. If the sum equals the original number, it is an
Armstrong number.

**Input**

A single integer N.

Print Armstrong Number or Not Armstrong Number.

# 36. Find the LCM of Two Numbers

**Problem Statement**

Given two integers A and B, find their Least Common Multiple (LCM).

**Input**

Two integers A and B.

**Output**

Print the LCM.

```python
a, b = map(int,input().strip().split())

def lcm(x,y):
    while y:
        x,y = y, x%y
    return x

print((a*b)//lcm(a,b))
```
