# 2. Check if a String is Palindrome

**Problem Statement**

Given a string, check whether it is a palindrome. A string is a palindrome if it reads the same forward and backward.

**Input**

A string S.

**Output**

Print Palindrome if S is a palindrome; otherwise print Not Palindrome.

**Examples**
- Example 1: `Input: ABCDCBA` -> `Output: Palindrome`
- Example 2: `Input: TAKE U FORWARD` -> `Output: Not Palindrome`

```python
n = input().lstrip()
print("Palindrome" if n[::-1] == n else "Not Palindrome")
```

# 3. Count Vowels, Consonants and Spaces

**Problem Statement**

Given a string, count the number of vowels, consonants, and spaces in it.

**Input**

A string S.

**Output**

Print the count of vowels, consonants, and spaces.

**Examples**
- Example 1: `Input: Take u forward is Awesome` -> `Output: Vowels: 10, Consonants: 11, White spaces: 4`
- Example 2: `Input: India won the cricket match` -> `Output: Vowels: 8, Consonants: 15, White spaces: 4`

```python
n = input()
v, c, s = 0,0,0
for i in n:
        if i in "aeiouAEIOU":
            v += 1
        elif i == " ":
            s += 1
        else:
            c += 1

print(f"Vowels: {v}, Consonants:{c}, White spaces: {s}")
```

# 9. Sum of the Numbers in a String

**Problem Statement**

**Input**

A string S containing letters and digits.

**Output**

Print the sum of all numbers present in S.

**Examples**
- Example 1: `Input: 123xyz` -> `Output: 123`
- Example 2: `Input: 1xyz23` -> `Output: 24`

```python
n = input().strip()
total = 0
dg = ""
for i in n:
    if i.isdigit():
        dg += i
    else:
        if dg == "":
            continue
        total += int(dg)
        dg = ""

if dg:
    total += int(dg)

print(total)
```

# 18. Change Every Letter to the Next Alphabet

**Problem Statement**

Replace every alphabetic character with the next letter in the alphabet. After z, continue with a; preserve letter case.

**Input**

A string S.

**Output**

Print the transformed string.

**Examples**
- Example 1: `Input: abcdxyz` -> `Output: bcdeyza`
- Example 2: `Input: Java` -> `Output: Kbwb`

```python
n = input().strip()
j = ""
for i in n:
    if i == "z":
        j += chr(97)
    elif i == "Z":
        j += chr(65)
    else:
        j += chr(ord(i)+1)

print(j)
```

# 22. Find the Word with Highest Repeated-Letter Frequency

**Problem Statement**

Find the word containing a letter with the highest frequency among all words. If no word contains a repeated letter, print -1.

**Input**

A string containing words separated by spaces.

**Output**

Print the first word with the highest repeated-letter frequency, or -1.

**Examples**
- Example 1: `Input: abcdefghij google microsoft` -> `Output: google`
- Example 2: `Input: cameron blue` -> `Output: -1`

```python
n = input().strip().split()
x = -1
mx = 1
for i in n:
    count = max(i.count(ch) for ch in i)
    if count > mx:
        x = i
        mx = count

print(x)
```
